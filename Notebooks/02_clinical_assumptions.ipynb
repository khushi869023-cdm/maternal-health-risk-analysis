This notebook defines the clinical and biological assumptions used to validate maternal health data.
These assumptions are translated into explicit validation rules that guide all downstream record-level and analytical checks.
No data validation or transformation is performed in this notebook.


# Clinical Variables under consideration 
 - Maternal age 
 - Systolic BP 
 - Diastolic BP
 - Body temprature
 - Heart rate 
 - Risk level 


# Logical and relational assumptions 

- Systolic blood pressure must be greater than or equal to diastolic blood pressure

- Extremely high blood pressure should not be labeled as low risk

- Very young maternal age combined with high-risk indicators requires review

# Risk level interpretation assumptions 

- Low, Mid, High are assumed to represent increasing clinical concern

- Risk level is treated as descriptive, not diagnostic

- Inconsistencies between vitals and risk level are flagged, not corrected

# Error vs Warning taxonomy 

-Error
- Biologically impossible
- Violates physical or physiological limits
- Indicates data entry or system failure

-Warning

- Clinically unlikely but possible
- Requires review, not removal
- May represent rare but valid cases

-Info

- Unusual patterns
- Monitored but not acted upon
